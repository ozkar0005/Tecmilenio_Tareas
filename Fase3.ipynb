To confirm the prediction functionality without a web page, you can create sample data and use the `service.score()` method directly in a code cell. This will print the predictions in your notebook output.

````python
import pandas as pd

# Prompt the user for input for each column
fecha_input = input("Ingrese la fecha para la que desea la predicción (YYYY-MM-DD): ")
bitcoin_price_input = float(input("Ingrese el último 'Bitcoin Precio Cierre' conocido: "))
interest_rate_input = float(input("Ingrese la última 'Tasa de interes promedio' conocida: "))
etf_price_input = float(input("Ingrese el último 'ETF precio cierre' conocido: "))

# Create sample data for scoring with user-provided values
# IMPORTANT: The 'Fecha' column is NOT included in scoring_data for the model input
# because the model expects only the feature columns. The date is handled internally
# by the time-series model for forecasting the next step.
scoring_data = pd.DataFrame({
    'Bitcoin Precio Cierre': [bitcoin_price_input],
    'Tasa de interes promedio': [interest_rate_input],
    'ETF precio cierre': [etf_price_input]
})

print("\n--- Su Entrada ---")
print(f"Fecha para la predicción: {fecha_input}")
print("Datos de características ingresados:")
display(scoring_data)
````

````python
# Make a prediction using the deployed web service
predictions = service.score(scoring_data)

print("\n--- Predicciones del Servicio Web ---")
display(predictions)
print(f"Predicción de 'Bitcoin Precio Cierre' para {fecha_input}: {predictions['predictions'][0]['values'][0][0]}")
````
